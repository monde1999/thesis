```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
def load_image(fn):
    img = cv2.imread(fn)
    img = img[:,:,0]
    print("image shape:", np.shape(img))
    plt.imshow(img)
    plt.show()
    return img
img = load_image("rabbit.png")
```
